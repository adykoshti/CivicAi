# Train GRAP Model for Ahmedabad

This notebook trains a Random Forest Classifier to predict the GRAP stage based on pollution levels.

```python
import pandas as pd
import numpy as np
from sklearn.ensemble import RandomForestClassifier
import joblib
import json
import os
```

```python
# Define features
FEATURES = ["PM2.5", "PM10", "NO2", "SO2", "CO", "Temperature", "Wind Speed"]

# Ahmedabad-adapted AQI breakpoints for GRAP
def get_grap_stage(aqi):
    if aqi <= 150:
        return 0  # Normal
    elif aqi <= 250:
        return 1  # Stage I
    elif aqi <= 350:
        return 2  # Stage II
    elif aqi <= 450:
        return 3  # Stage III
    else:
        return 4  # Stage IV
```

```python
# Load Dataset (Placeholder for actual Ahmedabad data)
# df = pd.read_csv("ahmedabad_pollution.csv")

# For demonstration, we generate synthetic data representing Ahmedabad conditions
np.random.seed(42)
n_samples = 1000

# Generate synthetic features
data = {
    "PM2.5": np.random.uniform(20, 300, n_samples),
    "PM10": np.random.uniform(50, 500, n_samples),  # Dust heavy
    "NO2": np.random.uniform(10, 100, n_samples),
    "SO2": np.random.uniform(5, 50, n_samples),
    "CO": np.random.uniform(0.1, 5.0, n_samples),
    "Temperature": np.random.uniform(10, 45, n_samples),
    "Wind Speed": np.random.uniform(0, 20, n_samples)
}
df = pd.DataFrame(data)

# Calculate AQI (Simplified approximation for training labels)
# In reality, use standard AQI calculation
df["AQI"] = np.maximum(df["PM2.5"] * 2, df["PM10"])

# Generate labels
df["grap_stage"] = df["AQI"].apply(get_grap_stage)

X = df[FEATURES]
y = df["grap_stage"]
```

```python
# Train Model
model = RandomForestClassifier(n_estimators=300, random_state=42)
model.fit(X, y)

# Save Model
os.makedirs("../models", exist_ok=True)
joblib.dump(model, "../models/grap_model.pkl")
print("Model saved to ../models/grap_model.pkl")
```
